```python
import pandas as pd

data = {
    'student': ['Asha', 'Babu', 'Chayan', 'Dipa', 'Emon', 'Farah'],
    'study_hours': [2, 5, 1, 7, 3, 6],
    'sleep_hours': [8, 6, 9, 5, 7, 6],
    'attendance': [70, 88, 55, 95, 75, 91],
    'snacks': [3, 2, 5, 1, 4, 2],
    'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}
df = pd.DataFrame(data)
# 16. Features and Label
print("16. Features & Label:")
print("Features: study_hours, sleep_hours, attendance, snacks")
print("Label: passed\n")
# 17. Dataset Dimensions
print("==17. Dataset Dimensions:==")
print(f"Total Samples: {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}\n")
# 18. Attendance >= 80
print("===18. Students with Attendance >= 80:==")
print(df[df['attendance'] >= 80].to_string(index=False))
print()
# 19. Average study hours (Passed)
print("===19. Average Study Hours for Passed Students:===")
avg_study = df[df['passed'] == 'Yes']['study_hours'].mean()
print(f"{avg_study:.1f} hours\n")
# 20. Sorted DataFrame
print("===20. Sorted by Attendance (Highest First):===")
print(df.sort_values(by='attendance', ascending=False).to_string(index=False))
print()
# 21. Causality Explanation
print("===21. Snack Purchases vs Failure:===")
print("No, correlation isn't causation. Six students is way too small a sample size, and students failed because of low study time and low attendance, not because they bought snacks.")
```

```python
import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)
    return np.random.randint(40, 101, size=5)

# Score outputs
print("Seeded run 1:", generate_scores(True))
print("Seeded run 2:", generate_scores(True))
print("Unseeded run 1:", generate_scores(False))
print("Unseeded run 2:", generate_scores(False))

# Observations and explanations
print("\nIdentical Outputs:")
print("Seeded run 1 and Seeded run 2 produced the exact same results.")

print("\nCan we fairly compare the models?")
print("No. Different random splits introduce variance caused by data luck—such as getting an easier test set—rather than actual model performance.")
print("For a fair comparison, both groups should use the same fixed seed or use K-Fold Cross-Validation.")
```
